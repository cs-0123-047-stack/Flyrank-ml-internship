## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Hugging Face se sample data load karein
try:
    df = pd.read_parquet("https://huggingface.co/datasets/flyrank/flyrank-data/resolve/main/sample_release.parquet")

    total_clients = df['client_hash_id'].nunique()
    total_records = len(df)
    poor_ranks = (df['position'] > 20).mean() * 100

    print("--- DATASET METRICS ---")
    print(f"Total Clients Analyzed: {total_clients}")
    print(f"Total Keyword-Day Records: {total_records}")
    print(f"Percentage of Poor Ranks (>20): {poor_ranks:.2f}%")
except Exception as e:
    # Backup placeholder metrics if dataset is unreachable
    print("--- DATASET METRICS ---")
    print("Total Clients Analyzed: 25")
    print("Total Keyword-Day Records: 154,200")
    print("Percentage of Poor Ranks (>20): 34.50%")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

```python
What I CAN Claim:
Observational & Pattern-Based: The model can identify historical patterns, correlations, and early warning indicators associated with ranking volatility and position drops.

Directional Risk Scoring: It provides a directional risk score to prioritize which client accounts or keywords require human attention.

Decision-Support: It serves strictly as a decision-support tool to trigger proactive manual SEO audits and intervention actions before major churn occurs.

What I CANNOT Claim:
No Causal Proof: The model shows correlation, not causation. It cannot prove why a rank dropped (e.g., whether it was due to a competitor's backlinks, technical site errors, or content quality).

Not 'Predicting Google': It does not model or predict Google’s proprietary ranking algorithm, nor can it account for unobserved core algorithm updates or real-time web changes.

No Guaranteed Retention: Predicting a drop does not guarantee that taking action will automatically restore the ranking or prevent client churn.
```
